## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

## 2. Load Dataset

```python
df = pd.read_csv("amazon.csv")
df.head()
```

## 3. Initial Data Inspection

```python
print("Initial Shape:", df.shape)
df.info()
df.isnull().sum()
```

## 4. Handle Missing Values

```python
# Fill missing review counts with 0
df['rating_count'] =df['rating_count'].fillna('0')
```

## 5. Clean Numerical Columns

```python
# Remove currency symbols, commas, and percent sign

df['discounted_price'] = df['discounted_price'].astype(str).str.replace('₹', '').str.replace(',', '')

df['actual_price'] = df['actual_price'].astype(str).str.replace('₹', '').str.replace(',', '')

df['discount_percentage']= df['discount_percentage'].astype(str).str.replace('%','')

df['rating_count']= df['rating_count'].astype(str).str.replace(',','')
```

## 6. Convert Columns to Numeric Format

```python
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')


df['discounted_price'] = pd.to_numeric(df['discounted_price'], errors='coerce')

df['actual_price'] = pd.to_numeric(df['actual_price'], errors='coerce')

df['discount_percentage'] = pd.to_numeric(df['discount_percentage'], errors='coerce')

df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce')
```

```python
df.describe()
```

## 7. Remove Critical Missing Values , Drop duplicate values

```python
df.dropna(subset=['discounted_price', 'actual_price', 'rating'], inplace=True)
df.drop_duplicates(inplace=True)
```

## 8. Feature Engineering

```python
# Main Category Extraction
df['main_category'] = df['category'].apply(lambda x: x.split('|')[0])
```

```python
# Savings Column
df['savings'] = df['actual_price'] - df['discounted_price']
```

```python
# Price Segmentation
df['price_range'] = pd.cut(
    df['discounted_price'],
    bins=[0,500,2000,5000,10000,50000],
    labels=['Budget','Affordable','Mid-Range','Premium','Luxury']
)
```

## 9. Final Dataset Validation

```python
print("Final Shape:", df.shape)
df.info()
df.describe()
df.isnull().sum()
```

## 10. Export Cleaned Dataset

```python
df.to_csv("amazon_cleaned.csv", index=False , encoding = "utf-8")
```
