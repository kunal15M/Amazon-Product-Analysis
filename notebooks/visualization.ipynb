# Amazon Product Pricing, Discounts & Customer Review Analysis

## Visualization Phase (Exploratory Data Analysis)

This notebook focuses on analyzing cleaned Amazon product data through visual exploration to uncover:
- Product category trends
- Pricing segmentation
- Discount strategies
- Customer satisfaction behavior
- Product value opportunities
- Business recommendations

# 1. Import Required Libraries
In this section, we import Python libraries required for:
- Data handling
- Visualization
- Statistical analysis

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2. Load Cleaned Dataset
Load the cleaned Amazon dataset prepared during preprocessing for visualization and business intelligence analysis.

```python
df = pd.read_csv("amazon_cleaned.csv")
```

# 3. Visualization Styling
Set global chart styles for professional and consistent visual presentation.

```python
sns.set(style="whitegrid")
plt.figure(figsize=(12,6))
```

# 4. Top Product Categories Analysis
## Objective:
Identify the most dominant product categories on Amazon.

### Business Value:
This helps reveal:
- Market concentration
- Competitive segments
- Product saturation

```python
category_counts = df['main_category'].value_counts()

sns.barplot(
    x=category_counts.values,
    y=category_counts.index
)

plt.title("Top 10 Product Categories")
plt.xlabel("Number of Products")
plt.ylabel("Category")
plt.show()
```

# 5. Average Discount by Category
## Objective:
Compare promotional strategies across major categories.

### Business Value:
Helps determine:
- Aggressive discounting sectors
- Premium vs budget strategies

```python
avg_discount = df.groupby('main_category')['discount_percentage'].mean().sort_values(ascending=False).head(10)

avg_discount.plot(kind='bar')
plt.title("Average Discount Percentage by Category")
plt.ylabel("Discount %")
plt.show()
```

# 6. Product Rating Distribution
## Objective:
Analyze customer satisfaction trends across products.

### Business Value:
- Understand overall product quality
- Detect trust patterns

```python
sns.histplot(df['rating'], bins=20, kde=True)

plt.title("Distribution of Product Ratings")
plt.xlabel("Rating")
plt.show()
```

# 7. Price Range Distribution
## Objective:
Segment products into pricing tiers:
- Budget
- Affordable
- Mid-Range
- Premium
- Luxury

### Business Value:
Helps understand Amazon’s product market structure.

```python
sns.countplot(x='price_range', data=df)

plt.title("Product Distribution by Price Range")
plt.xlabel("Price Range")
plt.ylabel("Count")
plt.show()
```

# 8. Top Products by Savings
## Objective:
Identify products offering maximum customer savings.

### Business Value:
- High perceived value
- Potential promotional opportunities

```python
top_savings = df.nlargest(10, 'savings')

sns.barplot(
    x='savings',
    y='product_name',
    data=top_savings
)

plt.title("Top 10 Products by Savings")
plt.xlabel("Savings")
plt.show()
```

# 9. Discount vs Rating Relationship
## Objective:
Evaluate whether deeper discounts influence customer ratings.

### Business Value:
Determine if discounting improves perceived product quality.

```python
sns.scatterplot(
    x='discount_percentage',
    y='rating',
    data=df
)

plt.title("Discount Percentage vs Rating")
plt.xlabel("Discount %")
plt.ylabel("Rating")
plt.show()
```

# 10. Correlation Heatmap
## Objective:
Study relationships between:
- Pricing
- Discounts
- Ratings
- Reviews
- Savings

### Business Value:
Supports strategic business intelligence.

```python
numeric_cols = [
    'discounted_price',
    'actual_price',
    'discount_percentage',
    'rating',
    'rating_count',
    'savings'
]

sns.heatmap(
    df[numeric_cols].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix")
plt.show()
```

# 11. Key Business Insights

## Pricing Insights:
- Budget and affordable products dominate
- Mid-range products often balance quality and affordability
- Luxury products offer larger nominal savings

## Discount Insights:
- Higher discounts do not guarantee stronger ratings
- Category discount strategies differ significantly

## Customer Behavior:
- Highly reviewed products often balance affordability and trust
- Ratings cluster in moderate-to-high satisfaction zones

# 12. Business Recommendations

### Strategic Recommendations:
